### 주요 해설 포인트

**리스트 기본 개념**
- 리스트는 순서가 있고 변경 가능한 컬렉션
- 0부터 시작하는 인덱싱 사용
- 음수 인덱싱으로 뒤에서부터 접근 가능
- 중복된 값 저장 가능, 다양한 타입 혼합 가능

**리스트 슬라이싱**
- `list[시작:끝:단계]` 형태로 사용
- 끝 인덱스는 포함되지 않음 (exclusive)
- 시작이나 끝을 생략하면 처음/끝까지 의미
- 음수 인덱스도 슬라이싱에서 사용 가능

**리스트 메소드**
- `append()`: 끝에 요소 추가
- `insert(index, value)`: 특정 위치에 삽입
- `remove(value)`: 값으로 요소 제거 (첫 번째만)
- `pop(index)`: 인덱스로 제거하고 값 반환
- `index(value)`: 값의 인덱스 찾기
- `count(value)`: 값의 개수 세기

**리스트 컴프리헨션**
- 한 줄로 새로운 리스트 생성하는 방법
- `[표현식 for 항목 in 반복가능객체 if 조건]`
- 간단한 변환에는 유용하지만, 복잡하면 가독성 저하
- 성능상 일반 반복문보다 빠름

**튜플 기본 개념**
- 순서가 있고 변경 불가능한 컬렉션
- 괄호 `()` 사용하여 생성 (생략 가능)
- 한 개 요소 튜플은 쉼표 필수: `(42,)`
- 리스트보다 메모리 효율적이고 빠름

**튜플 언패킹**
- 튜플의 값들을 여러 변수에 동시 할당
- `x, y = (10, 20)` 또는 `x, y = 10, 20`
- 변수 값 교환에 유용: `a, b = b, a`
- 함수에서 여러 값 반환할 때 활용

**리스트 vs 튜플**
- 가변성: 리스트(변경 가능) vs 튜플(변경 불가)
- 성능: 튜플이 더 빠르고 메모리 효율적
- 사용 사례: 변하는 데이터(리스트) vs 고정 데이터(튜플)
- 메소드: 리스트는 많은 메소드, 튜플은 제한적
