### 주요 해설 포인트

**파일 입출력 기본 개념**
- 파일은 데이터를 영구적으로 저장하는 수단
- 프로그램 종료 후에도 데이터가 보존됨
- `open()` 함수로 파일 열기, `close()` 메소드로 파일 닫기
- 적절한 파일 모드 선택이 중요

**파일 모드의 특성**
- `"r"` (읽기): 기존 파일 읽기, 파일이 없으면 FileNotFoundError
- `"w"` (쓰기): 새 파일 생성 또는 기존 파일 내용 완전 삭제 후 쓰기
- `"a"` (추가): 파일 끝에 내용 추가, 파일이 없으면 새로 생성
- 각 모드는 서로 다른 파일 포인터 위치를 가짐

**파일 읽기 메소드**
- `read()`: 전체 파일을 하나의 문자열로 읽기
- `readline()`: 한 줄씩 읽기, 줄바꿈 문자 포함
- `readlines()`: 모든 줄을 리스트로 읽기, 각 요소에 줄바꿈 문자 포함
- 대용량 파일 처리 시 메모리 사용량 고려 필요

**파일 쓰기 메소드**
- `write()`: 문자열 쓰기, 줄바꿈은 명시적으로 추가 필요
- `writelines()`: 문자열 리스트 쓰기, 자동 줄바꿈 없음
- 쓰기 모드에서는 데이터가 즉시 반영되지 않을 수 있음 (버퍼링)

**with문의 장점**
- 자동 파일 닫기로 리소스 누수 방지
- 예외 발생 시에도 안전한 파일 닫기 보장
- 코드의 가독성과 안전성 향상
- 컨텍스트 매니저 프로토콜 활용

**인코딩 처리**
- UTF-8 인코딩으로 유니코드 문자 (한글 등) 처리
- 인코딩 미지정 시 시스템 기본 인코딩 사용
- 문자 깨짐 방지를 위해 명시적 인코딩 지정 권장

**예외 처리**
- `FileNotFoundError`: 파일이 존재하지 않을 때
- `PermissionError`: 파일 접근 권한이 없을 때
- `UnicodeDecodeError`: 인코딩 문제가 있을 때
- try-except 문으로 적절한 예외 처리 필요

**성능 및 보안 고려사항**
- 대용량 파일은 한 번에 읽지 말고 청크 단위로 처리
- 파일 경로 검증으로 디렉터리 트래버셜 공격 방지
- 사용자 입력으로 파일명 생성 시 보안 검증 필요
- 파일 백업 및 버전 관리 고려

**실용적 응용**
- 설정 파일 관리
- 로그 시스템 구현
- 데이터 저장 및 분석
- 간단한 데이터베이스 대체
