### 주요 해설 포인트

**딕셔너리 기본 개념**
- 딕셔너리는 키-값 쌍으로 구성된 컬렉션
- 순서는 Python 3.7+ 에서 보장, 키를 통한 빠른 검색 가능
- 키는 고유해야 하며, 불변 객체만 사용 가능
- 값은 어떤 타입이든 사용 가능 (변경 가능)

**딕셔너리 접근과 조작**
- `dict[key]`: 직접 접근 (키가 없으면 KeyError)
- `dict.get(key, default)`: 안전한 접근 (키가 없으면 기본값 반환)
- `dict[key] = value`: 새 항목 추가 또는 기존 항목 수정
- `del dict[key]` 또는 `dict.pop(key)`: 항목 제거

**딕셔너리 메소드**
- `keys()`: 모든 키를 반환 (dict_keys 객체)
- `values()`: 모든 값을 반환 (dict_values 객체)
- `items()`: 키-값 쌍을 반환 (dict_items 객체)
- `get(key, default)`: 안전한 값 접근
- `pop(key)`: 키를 제거하고 값 반환
- `update(other)`: 다른 딕셔너리로 업데이트

**세트(Set) 특성**
- 중복을 허용하지 않는 컬렉션
- 순서가 없음 (Python 3.7+에서는 삽입 순서 유지)
- 중복 제거와 집합 연산에 유용
- `&` (교집합), `|` (합집합), `-` (차집합) 연산 지원

**데이터 구조 선택 기준**
- **리스트**: 순서가 중요하고 인덱스 접근이 필요할 때
- **튜플**: 불변 데이터, 좌표나 설정값 저장
- **딕셔너리**: 키-값 매핑, 빠른 검색이 필요할 때
- **세트**: 고유 항목만 필요하거나 집합 연산이 필요할 때

**성능 고려사항**
- 딕셔너리는 해시 테이블 기반으로 O(1) 평균 접근 시간
- 대용량 데이터에서 리스트 검색보다 딕셔너리가 훨씬 빠름
- 메모리 사용량은 리스트보다 딕셔너리가 더 많음

**모범 사례**
- 의미 있는 키 이름 사용
- `get()` 메소드로 안전한 접근
- 키 존재 여부 확인 후 접근
- 적절한 데이터 구조 선택
