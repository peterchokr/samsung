### 주요 해설 포인트

**입출력 기본 개념**
- 입력: 사용자로부터 데이터를 받는 과정
- 출력: 프로그램이 사용자에게 결과를 보여주는 과정
- 대화형 프로그램의 핵심 요소

**input() 함수 특징**
- 항상 문자열(str) 타입으로 반환
- 사용자가 Enter를 누를 때까지 대기
- 수학 계산을 위해서는 타입 변환 필수

**print() 함수 고급 기능**
- sep: 여러 값 사이의 구분자 설정
- end: 출력 후 나타날 문자 설정
- f-string: 변수와 표현식을 문자열에 삽입

**데이터 타입 변환**
- int(): 문자열을 정수로 변환
- float(): 문자열을 실수로 변환
- str(): 숫자를 문자열로 변환

**사용자 경험 향상**
- 명확한 프롬프트 메시지 제공
- 예시와 함께 입력 형식 안내
- 오류 상황에 대한 적절한 처리
