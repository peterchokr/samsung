### 주요 해설 포인트

**반복문 기본 개념**
- 같은 코드를 반복 실행하여 효율성 향상
- 코드 재사용성과 유지보수성 개선
- 대용량 데이터 처리에 필수적

**for문 특징**
- 정해진 횟수나 시퀀스 처리에 적합
- range() 함수로 숫자 시퀀스 생성
- 문자열, 리스트 등 반복 가능한 객체 처리

**range() 함수**
- range(stop): 0부터 stop-1까지
- range(start, stop): start부터 stop-1까지
- range(start, stop, step): 시작점, 끝점, 증가값 지정

**while문 특징**
- 조건이 True인 동안 반복 실행
- 반복 횟수를 모를 때 사용
- 조건 변수 업데이트 필수 (무한 루프 방지)

**제어문**
- break: 반복문 완전 종료
- continue: 현재 반복 건너뛰고 다음 반복으로

**for vs while 선택 기준**
- for: 횟수가 정해진 반복, 시퀀스 처리
- while: 조건 기반 반복, 사용자 상호작용
