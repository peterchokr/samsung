### 13번
다음 코드의 결과는?

````
def outer_function():
    def inner_function():
        return "내부 함수"
    return inner_function()

print(outer_function())
````

1) outer_function
2) inner_function
3) 내부 함수
4) 오류 발생

2) return [a, b, c]
3) return (a, b, c)
4) 위의 모든 방법이 가능

### 15번
다음 코드의 실행 결과로 옳은 것은?

````
count = 0

def increment():
    global count
    count += 1
    return count

print(increment())
print(increment())
````

1) 0, 0
2) 1, 1
3) 1, 2
4) 2, 2
