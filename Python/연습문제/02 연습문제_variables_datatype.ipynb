### 주요 해설 포인트

**변수 기본 개념**
- 변수는 데이터를 저장하는 메모리 공간의 이름
- 라벨이 붙은 상자에 비유할 수 있음
- 값을 저장, 참조, 변경 가능

**변수 명명 규칙**
- 문자나 밑줄로 시작 (숫자 시작 불가)
- 문자, 숫자, 밑줄만 사용 (특수문자 사용 불가)
- 파이썬 키워드 사용 금지 (if, for, class 등)
- 대소문자 구분 (Name과 name은 다른 변수)

**기본 데이터 타입**
- int: 정수 (나이, 개수 등)
- float: 실수 (키, 몸무게, 가격 등)
- str: 문자열 (이름, 주소 등)
- bool: 불린 (True, False만 가능)

**type() 함수**
- 변수의 데이터 타입 확인
- 디버깅과 코드 검증에 유용
- `<class 'type_name'>` 형태로 출력
