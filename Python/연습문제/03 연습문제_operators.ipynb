### 주요 해설 포인트

**산술 연산자**
- 기본 연산자: +, -, *, / (일반적인 수학 연산)
- 특수 연산자: // (정수 나눗셈), % (나머지), ** (거듭제곱)
- / 는 항상 float 결과, // 는 정수 결과 반환

**비교 연산자**
- 모든 비교 연산자는 불린(True/False) 값 반환
- == (같음), != (다름), <, >, <=, >= (크기 비교)
- 조건문과 논리 연산의 기초가 됨

**논리 연산자**
- and: 모든 조건이 True여야 True 반환
- or: 하나 이상의 조건이 True면 True 반환
- not: 조건을 반대로 변환

**할당 연산자**
- 복합 할당: +=, -=, *=, /= (연산과 할당을 동시에)
- 코드 간소화와 가독성 향상

**연산자 우선순위**
- PEMDAS 규칙: 괄호 → 지수 → 곱셈/나눗셈 → 덧셈/뺄셈
- 복잡한 표현식에서는 괄호로 명확성 확보
