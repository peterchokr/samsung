### 주요 해설 포인트

**문자열 기본 개념**
- 문자열은 따옴표로 둘러싸인 문자들의 연속
- 불변(immutable) 특성: 생성 후 변경 불가
- 인덱싱은 0부터 시작, 음수 인덱스는 뒤에서부터

**인덱싱과 슬라이싱**
- 양수 인덱스: 0, 1, 2... (앞에서부터)
- 음수 인덱스: -1, -2, -3... (뒤에서부터)
- 슬라이싱: [start:end:step], end는 포함되지 않음

**문자열 메소드**
- 대소문자 변환: .upper(), .lower(), .title()
- 공백 처리: .strip(), .lstrip(), .rstrip()
- 검색: .find(), .count(), .startswith(), .endswith()
- 교체: .replace()

**문자열 형식화**
- f-string: f"텍스트 {변수}" (권장 방법)
- .format(): "텍스트 {}".format(변수)
- 표현식과 메소드 호출 가능
