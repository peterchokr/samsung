### 주요 해설 포인트

**클래스와 객체 기본 개념**
- 클래스는 객체를 만들기 위한 설계도나 템플릿
- 객체(인스턴스)는 클래스로부터 생성된 실체
- 하나의 클래스로 여러 개의 객체 생성 가능
- 실세계의 사물이나 개념을 모델링하는 도구

**클래스 정의와 객체 생성**
- `class ClassName:` 형태로 클래스 정의
- `object = ClassName()` 형태로 객체 생성
- 클래스명은 일반적으로 PascalCase 사용
- `pass` 키워드로 빈 클래스 정의 가능

**속성과 메소드**
- 속성(Attribute): 객체의 상태를 나타내는 변수
- 메소드(Method): 객체의 행동을 정의하는 함수
- 인스턴스 변수: 각 객체마다 독립적으로 존재
- 클래스 변수: 모든 객체가 공유하는 변수

**self 매개변수**
- 메소드의 첫 번째 매개변수로 반드시 포함
- 현재 객체(인스턴스) 자신을 가리킴
- 메소드 내에서 객체의 속성과 메소드에 접근할 때 사용
- 호출 시에는 자동으로 전달되므로 명시적으로 전달하지 않음

**__init__ 메소드**
- 객체가 생성될 때 자동으로 호출되는 특수 메소드
- 생성자(constructor) 역할 수행
- 객체의 초기 상태를 설정하는 데 사용
- 매개변수를 통해 초기값을 받아 속성에 할당

**상속(Inheritance)**
- 기존 클래스(부모)의 속성과 메소드를 새 클래스(자식)가 물려받는 것
- `class ChildClass(ParentClass):` 형태로 정의
- 코드 재사용성 향상 및 계층적 구조 형성
- `super()` 함수로 부모 클래스의 메소드 호출 가능

**메소드 오버라이딩**
- 자식 클래스에서 부모 클래스의 메소드를 재정의하는 것
- 같은 이름의 메소드를 자식 클래스에서 새로 구현
- 다형성(Polymorphism)의 기초가 됨
- 자식 클래스의 특성에 맞게 기능 변경 가능

**객체지향 프로그래밍의 장점**
- 코드 재사용성: 클래스와 상속을 통한 코드 재활용
- 유지보수성: 모듈화된 구조로 수정과 확장 용이
- 가독성: 실세계 모델링으로 이해하기 쉬운 코드
- 확장성: 상속과 다형성을 통한 기능 확장

**캡슐화(Encapsulation)**
- 데이터와 메소드를 하나의 단위로 묶는 것
- 외부에서 직접 접근을 제한하여 데이터 보호
- 파이썬에서는 언더스코어(_)를 이용한 접근 제어
- 정보 은닉(Information Hiding) 원칙 구현

**모범 사례**
- 클래스명은 명사형으로, 메소드명은 동사형으로 명명
- 하나의 클래스는 하나의 책임만 가져야 함 (단일 책임 원칙)
- 적절한 수준의 캡슐화로 데이터 보호
- 상속보다는 조합(Composition)을 우선 고려

**주의사항**
- 메소드에서 self 매개변수 누락하지 않기
- 클래스 변수와 인스턴스 변수 구분하기
- 과도한 상속 깊이 피하기
- 적절한 추상화 수준 유지하기
