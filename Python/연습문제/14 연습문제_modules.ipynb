### 주요 해설 포인트

**모듈 기본 개념**
- 모듈은 다른 프로그램에서 재사용할 수 있는 파이썬 코드 파일
- 함수, 변수, 클래스 등을 포함할 수 있음
- 코드 재사용성, 구조화, 유지보수성 향상
- `.py` 확장자를 가진 파이썬 파일이 모듈이 됨

**import 문의 다양한 형태**
- `import 모듈명`: 전체 모듈을 가져와 `모듈명.함수명()` 형태로 사용
- `from 모듈명 import 함수명`: 특정 함수만 가져와 직접 사용
- `import 모듈명 as 별칭`: 모듈에 짧은 별칭을 부여
- `from 모듈명 import *`: 모든 함수를 가져와 직접 사용 (권장하지 않음)

**표준 라이브러리 주요 모듈**
- **math 모듈**: 수학 함수와 상수 제공
  - `sqrt()`: 제곱근, `pow()`: 거듭제곱, `ceil()`: 올림, `floor()`: 내림
  - `pi`: 원주율, `e`: 자연상수
- **random 모듈**: 난수 생성과 무작위 선택
  - `randint(a, b)`: 정수 난수, `random()`: 0~1 실수 난수
  - `choice()`: 리스트에서 무작위 선택, `shuffle()`: 리스트 섞기
- **datetime 모듈**: 날짜와 시간 처리
  - `datetime.now()`: 현재 날짜/시간, `date.today()`: 오늘 날짜
  - `strftime()`: 날짜 형식 지정

**모듈 작성 모범 사례**
- 명확하고 설명적인 함수명 사용
- 함수에 docstring으로 문서화 추가
- 관련 기능들을 논리적으로 그룹화
- 모듈 레벨에서 실행되는 코드 최소화
- `if __name__ == "__main__":` 패턴 활용

**네임스페이스와 이름 충돌**
- 각 모듈은 독립적인 네임스페이스를 가짐
- `from module import *` 사용 시 이름 충돌 위험
- 별칭 사용으로 충돌 방지 가능
- 명시적 import가 더 안전하고 명확

**모듈의 종류**
- **내장 모듈**: 파이썬 인터프리터에 포함 (예: `builtins`)
- **표준 라이브러리**: 파이썬 설치 시 포함 (예: `math`, `random`)
- **서드파티 모듈**: 외부에서 설치 (예: `numpy`, `pandas`)
- **사용자 정의 모듈**: 직접 작성한 모듈

**모듈러 프로그래밍의 장점**
- 코드 재사용성: 한 번 작성하여 여러 곳에서 사용
- 유지보수성: 기능별로 분리되어 수정이 용이
- 테스트 용이성: 각 모듈을 독립적으로 테스트 가능
- 협업 효율성: 팀원들이 각각 다른 모듈 개발 가능

**import 문 작성 위치와 관례**
- 파일 맨 위에 작성 (주석 제외)
- 표준 라이브러리 → 서드파티 → 사용자 정의 순서
- 각 그룹 사이에 빈 줄로 구분
- 함수 내부에서의 import는 특별한 경우에만 사용

**주의사항과 안티패턴**
- `from module import *` 남용 지양
- 순환 import 문제 주의
- 모듈명과 변수명 충돌 방지
- 불필요한 모듈 import 지양
- 모듈 내 전역 변수 남용 피하기
