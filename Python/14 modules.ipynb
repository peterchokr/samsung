## 📦 1. 모듈이란 무엇인가?

### 모듈의 개념

**모듈(Module)**은 다른 프로그램에서 사용할 수 있는 파이썬 코드가 포함된 파일입니다. 모듈을 **도구상자**처럼 생각해보세요 - 각각 필요할 때 사용할 수 있는 특정 도구(함수)들이 들어있습니다.

### 모듈을 사용하는 이유

#### 1. 코드 재사용성 (Code Reusability)

같은 함수를 반복해서 작성하는 대신, 모듈에 저장하여 여러 프로그램에서 사용할 수 있습니다.

#### 2. 코드 구성 (Code Organization)

모듈은 관련된 함수들을 함께 그룹화하여 코드를 체계적으로 유지하는 데 도움이 됩니다.

#### 3. 협업 (Collaboration)

다른 프로그래머들이 서로 다른 모듈에서 작업한 후 이를 결합할 수 있습니다.

### 실생활 비유

모듈은 마트의 **각 코너**와 같습니다:

```
수학 관련 코너 (math 모듈):
- 계산기, 공학계산기, 각도기
- 계산 관련 모든 도구들

게임 코너 (random 모듈):
- 주사위, 카드게임, 복권
- 무작위성이 필요한 모든 것들

시계 코너 (datetime 모듈):
- 시계, 달력, 타이머
- 시간 관련 모든 도구들
```

### 모듈의 종류

#### 1. 내장 모듈 (Built-in Modules)

파이썬과 함께 자동으로 제공되는 모듈들

#### 2. 표준 라이브러리 모듈 (Standard Library Modules)

파이썬 설치와 함께 포함된 추가 모듈들

#### 3. 서드파티 모듈 (Third-party Modules)

다른 개발자들이 만든 설치 가능한 모듈들

#### 4. 사용자 정의 모듈 (Custom Modules)

직접 만드는 모듈들

## 📚 3. 표준 라이브러리 소개

### math 모듈 (수학 모듈)

**math** 모듈은 수학 함수와 상수들을 제공합니다.

#### 주요 math 함수들

````python
import math

print("=== MATH 모듈 예시 ===")

# 기본 함수들
print(f"16의 제곱근: {math.sqrt(16)}")
print(f"2의 3제곱: {math.pow(2, 3)}")
print(f"4.3의 올림: {math.ceil(4.3)}")
print(f"4.7의 내림: {math.floor(4.7)}")

# 상수들
print(f"파이 값: {math.pi}")
print(f"자연상수 e: {math.e}")

# 삼각함수 (도를 라디안으로 먼저 변환)
angle_degree = 90
angle_radian = math.radians(angle_degree)
print(f"90도의 사인값: {math.sin(angle_radian)}")
````

### random 모듈 (랜덤 모듈)

**random** 모듈은 난수를 생성하고 무작위 선택을 합니다.

#### 주요 random 함수들

````python
import random

print("=== RANDOM 모듈 예시 ===")

# 랜덤 정수
print(f"1-10 사이 랜덤 정수: {random.randint(1, 10)}")
print(f"1-100 사이 랜덤 정수: {random.randint(1, 100)}")

# 랜덤 실수
print(f"0-1 사이 랜덤 실수: {random.random()}")
print(f"1-10 사이 랜덤 실수: {random.uniform(1, 10)}")

# 랜덤 선택
colors = ["빨강", "파랑", "초록", "노랑"]
print(f"랜덤 색깔: {random.choice(colors)}")

# 리스트 섞기
numbers = [1, 2, 3, 4, 5]
random.shuffle(numbers)
print(f"섞인 n들: {numbers}")
````

### datetime 모듈 (날짜시간 모듈)

**datetime** 모듈은 날짜와 시간 작업에 도움을 줍니다.

#### 주요 datetime 함수들

````python
import datetime

print("=== DATETIME 모듈 예시 ===")

# 현재 날짜와 시간
now = datetime.datetime.now()
print(f"현재 날짜와 시간: {now}")

# today 날짜만
today = datetime.date.today()
print(f"오늘 날짜: {today}")

# 날짜 형식 지정
formated_date = now.strftime("%Y-%m-%d %H:%M:%S")
print(f"형식 지정된 날짜: {formated_date}")

# 간단한 형식
simple_date = now.strftime("%Y년 %m월 %d일")
print(f"간단한 형식: {simple_date}")

# 요일 name
my_day = now.strftime("%A")
print(f"오늘은: {my_day}")
````

### 실용적인 표준 라이브러리 활용 예시

````python
# 여러 모듈을 조합한 유용한 프로그램
import math
import random
import datetime

def math_quiz():
    """여러 모듈을 사용한 일일 수학 퀴즈 생성"""
  
    # 현재 날짜 가져오기
    today = datetime.date.today().strftime("%Y년 %m월 %d일")
    print(f"=== 일일 수학 퀴즈 - {today} ===")
  
    # 문제용 랜덤 n 생성
    n1 = random.randint(1, 20)
    n2 = random.randint(1, 20)
  
    print(f"문제 1: {n1} + {n2}는?")
    print(f"정답: {n1 + n2}")
  
    # 제곱근 문제
    number = random.choice([4, 9, 16, 25, 36, 49])
    print(f"문제 2: {number}의 제곱근은?")
    print(f"정답: {int(math.sqrt(number))}")
  
    # 원의 면적 문제
    radius = random.randint(2, 8)
    area = math.pi * radius * radius
    print(f"문제 3: 반지름이 {radius}인 원의 면적은?")
    print(f"정답: {area:.2f}")
  
    print("퀴즈 완료!")

# 퀴즈 실행
math_quiz()
````

## 🔨 4. 사용자 정의 모듈 만들기

### 사용자 정의 모듈이란?

**사용자 정의 모듈**은 다른 프로그램에서 사용할 수 있는 자신만의 함수와 변수로 만든 파이썬 파일입니다.

### 단계별 모듈 생성

#### 1단계: 모듈 파일 생성

`my_utilities.py`라는 파일을 만드세요:

````python
# my_utilities.py - 유용한 함수들의 모음

def greeting(name):
    """사용자에게 친근한 msg로 greeting"""
    return f"안녕하세요, {name}님! 우리 프로그램에 오신 것을 환영합니다!"

def calculate_average(numbers):
    """n 리스트의 average 계산"""
    if len(numbers) == 0:
        return 0
    return sum(numbers) / len(numbers)

def find_average(n):
    """숫자가 짝수인지 확인"""
    return n % 2 == 0

def find_initial(fullname):
    """전체 name에서 이니셜 추출"""
    name_list = fullname.split()
    initials = ""
    for name in name_list:
        initials += name[0].upper()
    return initials

# 모듈 상수
pi = 3.14159
greeting_msg = "유틸리티 모듈을 사용해 주셔서 감사합니다!"

# 모듈이 import될 때 실행되는 코드
print("my_utilities 모듈이 성공적으로 로드되었습니다!")
````

#### 2단계: 모듈 사용하기

`main_program.py`라는 메인 프로그램 파일을 만드세요:

````python
# main_program.py - 사용자 정의 모듈 사용

import my_utilities

print("=== 사용자 정의 모듈 테스트 ===")

# greeting 함수 테스트
msg = my_utilities.greeting("김철수")
print(msg)

# average 계산 테스트
scores = [85, 92, 78, 96, 88]
average = my_utilities.calculate_average(scores)
print(f"평균 점수: {average}")

# 짝수 확인 테스트
n = 42
if my_utilities.find_average(n):
    print(f"{n}는 짝수입니다")
else:
    print(f"{n}는 홀수입니다")

# 이니셜 추출 테스트
name = "김철수 박영희"
i = my_utilities.find_initial(name)
print(f"{name}의 이니셜: {i}")

# 모듈 상수 사용
print(f"모듈의 파이 값: {my_utilities.pi}")
print(my_utilities.greeting_msg)
````

### 실습 2: 랜덤 게임 프로그램

**문제**: random 모듈을 사용하여 간단한 n 맞히기 게임을 만드세요.

**해답**:

````python
# guessing_game.py - 간단한 n 맞히기 게임

import random

def guess_number():
    """간단한 n 맞히기 게임"""
    print("=== n 맞히기 게임 ===")
  
    # 컴퓨터가 1-10 사이의 랜덤 n 선택
    secret_n = random.randint(1, 10)
  
    print("1부터 10 사이의 n를 생각하고 있어요")
    guess = int(input("n를 맞춰보세요: "))
  
    if guess == secret_n:
        print(f"정답! n는 {secret_n}였습니다")
    else:
        print(f"틀렸어요! n는 {secret_n}였습니다")

def dice_game():
    """간단한 주사위 굴리기"""
    print("\n=== 주사위 게임 ===")
    input("엔터를 눌러 주사위를 굴리세요...")
  
    dice1 = random.randint(1, 6)
    dice2 = random.randint(1, 6)
    total = dice1 + dice2
  
    print(f"주사위 결과: {dice1}과 {dice2}")
    print(f"total: {total}")
  
    if total == 7:
        print("럭키 세븐! 당신이 이겼습니다!")
    else:
        print("다시 시도해보세요!")

# 게임 실행
guess_number()
dice_game()
````

### 실습 3: 날짜/시간 프로그램

**문제**: datetime 모듈을 사용하여 현재 날짜와 시간 정보를 보여주는 간단한 프로그램을 만드세요.

**해답**:

````python
# datetime_info.py - 간단한 날짜와 시간 프로그램

import datetime

def view_today():
    """현재 날짜와 시간 보기"""
    now = datetime.datetime.now()
    today = datetime.date.today()
  
    print("=== 현재 날짜 & 시간 ===")
    print(f"today 날짜: {today}")
    print(f"현재 시간: {now.strftime('%H:%M:%S')}")
    print(f"요일: {now.strftime('%A')}")
    print(f"월: {now.strftime('%B')}")

def calcuate_age(birthyear):
    """출생연도로 age 계산"""
    current_year = datetime.date.today().year
    age = current_year - birthyear
    print(f"\n{birthyear}년생이라면, 현재 나이는 {age}세입니다")

def date_to_newyear():
    """newyear_date까지 남은 날짜 계산"""
    today = datetime.date.today()
    nextyear = today.year + 1
    newyear_date = datetime.date(nextyear, 1, 1)
    remaining_days = (newyear_date - today).days
    print(f"\n신정까지 남은 날짜: {remaining_days}일")

# 함수들 실행
view_today()
calcuate_age(2000)
date_to_newyear()
````
